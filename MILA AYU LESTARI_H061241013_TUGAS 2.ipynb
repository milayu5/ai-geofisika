## **Eksekusi Pendukung dengan Data Sintetis**

Outputnya bersifat konseptual karena prompt melarang pembuatan nilai kuantitatif tanpa data. Untuk memperlihatkan bagaimana langkah analisis tersebut dikerjakan pada data spasial, bagian ini mengeksekusi simulasi sederhana menggunakan **NumPy** dan **Matplotlib**. Area model berukuran 300 m × 200 m dengan resolusi 1 m per piksel, berupa lereng yang menurun ke timur dan memuat zona longsoran, alur lembah lurus, serta tutupan vegetasi buatan.

Sel-sel berikut berurutan dan saling bergantung, sehingga sebaiknya dijalankan dari atas ke bawah.

### **1. Membangun DSM dan DTM Sintetis**

Data sintetis : grid permukaan dengan lereng dasar, zona longsoran, alur lembah lurus, dan kanopi vegetasi.

Untuk memperlihatkan perbedaan DSM (permukaan teratas) dan DTM (permukaan tanah), serta tinggi kanopi sebagai selisih keduanya.

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({"figure.dpi": 100, "axes.grid": False, "font.size": 10})
rng = np.random.default_rng(2026)   # seed tetap agar hasil dapat diulang

def smooth(arr, k=5):
    """Perataan sederhana (moving average) tanpa library tambahan."""
    ker = np.ones(k) / k
    out = np.apply_along_axis(lambda r: np.convolve(r, ker, mode="same"), 1, arr)
    out = np.apply_along_axis(lambda c: np.convolve(c, ker, mode="same"), 0, out)
    return out

# ------------------------------------------------------------
# 1. GRID PERMUKAAN SINTETIS (300 m x 200 m, resolusi 1 m/piksel)
# ------------------------------------------------------------
res = 1.0
x = np.arange(0, 300, res)
y = np.arange(0, 200, res)
X, Y = np.meshgrid(x, y)

# Lereng dasar menurun ke arah timur + undulasi alami
dtm_base = 120 - 0.28 * X + 0.02 * Y + 5 * np.sin(X / 45) * np.cos(Y / 38)

# Zona longsoran sintetis: gawir (kehilangan massa) di bagian atas lereng
r = np.sqrt(((X - 165) / 55) ** 2 + ((Y - 100) / 42) ** 2)
dtm_t1 = dtm_base - 7.0 * np.exp(-r ** 8)

# Kelurusan sintetis berupa alur lembah lurus (arah barat laut - tenggara) di bagian barat
jarak_garis = (Y - (60 - 0.5 * X)) / np.sqrt(1.25)
dtm_t1 = dtm_t1 - 1.3 * np.exp(-(jarak_garis / 3) ** 2)

# Tutupan vegetasi sintetis (tinggi kanopi) -> pembeda DSM dan DTM
canopy = np.zeros_like(X)
for _ in range(70):
    cx, cy = rng.uniform(0, 300), rng.uniform(0, 200)
    h, s = rng.uniform(5, 13), rng.uniform(5, 11)
    canopy += h * np.exp(-((X - cx) ** 2 + (Y - cy) ** 2) / (2 * s ** 2))
canopy = np.clip(canopy, 0, 16)

dtm = dtm_t1                                            # model permukaan tanah
dsm = dtm + canopy + rng.normal(0, 0.05, X.shape)       # model permukaan (tanah + vegetasi)
chm = dsm - dtm                                         # Canopy Height Model

print("DATA SINTETIS - GRID PERMUKAAN")
print(f"Ukuran grid        : {X.shape[1]} x {X.shape[0]} piksel ({x.max()+res:.0f} m x {y.max()+res:.0f} m)")
print(f"Elevasi DTM        : {dtm.min():.1f} - {dtm.max():.1f} m")
print(f"Elevasi DSM        : {dsm.min():.1f} - {dsm.max():.1f} m")
print(f"Tinggi kanopi maks : {chm.max():.1f} m")
print("Catatan: seluruh nilai dibuat untuk demonstrasi, bukan hasil survei lapangan.")

fig, ax = plt.subplots(1, 3, figsize=(15, 4.2))
for a, data, jd, cm, lb in zip(ax, [dsm, dtm, chm],
        ["DSM sintetis", "DTM sintetis", "Tinggi kanopi (DSM - DTM)"],
        ["terrain", "terrain", "Greens"], ["Elevasi (m)", "Elevasi (m)", "Tinggi (m)"]):
    im = a.imshow(data, origin="lower", extent=[0, 300, 0, 200], cmap=cm)
    a.set_title(jd); a.set_xlabel("X lokal (m)"); a.set_ylabel("Y lokal (m)")
    plt.colorbar(im, ax=a, label=lb, fraction=0.046)
plt.tight_layout()
plt.show()
```

**Interpretasi:** DSM memiliki rentang elevasi 38,1–131,5 m, sedikit lebih tinggi daripada DTM (37,6–124,9 m), dan selisihnya mencapai 16,2 m pada lokasi dengan kanopi paling tinggi. Selisih ini menunjukkan bahwa DSM merekam vegetasi, sehingga analisis bentuk lahan dan kemiringan lereng lebih tepat dilakukan pada DTM. Pada data drone nyata, DTM diperoleh dari klasifikasi titik tanah pada point cloud dan ketelitiannya menurun di bawah vegetasi yang rapat.

### **2. Perencanaan Penerbangan: GSD, Overlap, dan GCP**

Data asumsi : spesifikasi kamera 1 inci, tinggi terbang 100 m, overlap depan 80% dan overlap samping 70%.

Untuk menghitung GSD, jarak antarfoto, jarak antarjalur, dan perkiraan jumlah foto, serta menampilkan rencana jalur terbang dan sebaran GCP.

```python
# ------------------------------------------------------------
# PARAMETER KAMERA & PENERBANGAN (ASUMSI UNTUK DEMONSTRASI)
# ------------------------------------------------------------
lebar_sensor = 13.2e-3     # m   (asumsi sensor 1 inci)
tinggi_sensor = 8.8e-3     # m
fokus = 8.8e-3             # m
piksel_lebar, piksel_tinggi = 5472, 3648
tinggi_terbang = 100.0     # m di atas permukaan (asumsi)
overlap_depan, overlap_samping = 0.80, 0.70

# GSD = (lebar sensor x tinggi terbang) / (panjang fokus x jumlah piksel lebar)
gsd = (lebar_sensor * tinggi_terbang) / (fokus * piksel_lebar)        # m/piksel
foot_lebar = gsd * piksel_lebar                                        # m (tegak lurus jalur)
foot_tinggi = gsd * piksel_tinggi                                      # m (searah jalur)

jarak_foto = foot_tinggi * (1 - overlap_depan)                         # jarak antarfoto
jarak_jalur = foot_lebar * (1 - overlap_samping)                       # jarak antarjalur

panjang_area, lebar_area = 300.0, 200.0
n_jalur = int(np.ceil(lebar_area / jarak_jalur)) + 1
foto_per_jalur = int(np.ceil(panjang_area / jarak_foto)) + 1

print("PERENCANAAN PENERBANGAN - PARAMETER ASUMSI")
print(f"GSD                       : {gsd*100:.2f} cm/piksel")
print(f"Footprint satu foto       : {foot_lebar:.1f} m x {foot_tinggi:.1f} m")
print(f"Jarak antarfoto           : {jarak_foto:.1f} m (overlap depan {overlap_depan*100:.0f}%)")
print(f"Jarak antarjalur          : {jarak_jalur:.1f} m (overlap samping {overlap_samping*100:.0f}%)")
print(f"Perkiraan jumlah jalur    : {n_jalur}")
print(f"Perkiraan foto per jalur  : {foto_per_jalur}")
print(f"Perkiraan total foto      : {n_jalur * foto_per_jalur}")
print("Catatan: spesifikasi kamera dan tinggi terbang hanya asumsi demonstrasi.")

# Hubungan tinggi terbang dengan GSD
tinggi_uji = np.arange(30, 151, 10)
gsd_uji = (lebar_sensor * tinggi_uji) / (fokus * piksel_lebar) * 100

# Posisi jalur terbang dan GCP sintetis
y_jalur = np.linspace(0, lebar_area, n_jalur)
gcp = np.array([[15, 15], [285, 15], [150, 100], [15, 185], [285, 185]])

fig, ax = plt.subplots(1, 2, figsize=(13, 4.6))
ax[0].plot(tinggi_uji, gsd_uji, "o-", color="tab:blue")
ax[0].axvline(tinggi_terbang, ls="--", color="gray")
ax[0].set_xlabel("Tinggi terbang (m)"); ax[0].set_ylabel("GSD (cm/piksel)")
ax[0].set_title("Tinggi terbang vs GSD"); ax[0].grid(alpha=0.3)

for i, yy in enumerate(y_jalur):
    ax[1].plot([0, panjang_area], [yy, yy], color="tab:blue", lw=1,
               label="Jalur terbang" if i == 0 else None)
ax[1].scatter(gcp[:, 0], gcp[:, 1], marker="^", s=90, color="tab:red", zorder=3, label="GCP (sintetis)")
ax[1].set_xlim(-10, 310); ax[1].set_ylim(-10, 210)
ax[1].set_xlabel("X lokal (m)"); ax[1].set_ylabel("Y lokal (m)")
ax[1].set_title("Rencana jalur terbang dan sebaran GCP"); ax[1].legend(loc="upper right")
plt.tight_layout()
plt.show()
```

**Interpretasi:** Dengan parameter asumsi, GSD bernilai 2,74 cm/piksel dan satu foto mencakup sekitar 150 m × 100 m. Overlap yang tinggi membuat foto diambil tiap 20 m dan jalur terbang berjarak 45 m, sehingga area 300 m × 200 m memerlukan sekitar 6 jalur dan 102 foto. Grafik kiri menunjukkan GSD bertambah linear terhadap tinggi terbang, yaitu detail permukaan makin kasar pada ketinggian lebih besar. Lima GCP diletakkan di sudut dan tengah area agar mengikat model secara merata. Pada kawasan berlereng nyata, tinggi terbang perlu mengikuti topografi agar GSD tetap seragam.

### **3. Analisis Kemiringan, Arah Hadap, dan Hillshade**

Data sintetis : DTM hasil Eksekusi 1.

Untuk menurunkan peta morfometri dari DTM dan menghitung proporsi area pada tiap kelas kemiringan.

```python
# ------------------------------------------------------------
# TURUNAN MORFOMETRI DARI DTM: KEMIRINGAN, ARAH HADAP, HILLSHADE
# ------------------------------------------------------------
dtm_s = smooth(dtm, 3)[1:-1, 1:-1]            # buang tepi akibat efek perataan
Xc, Yc = X[1:-1, 1:-1], Y[1:-1, 1:-1]

dz_dy, dz_dx = np.gradient(dtm_s, res)
slope = np.degrees(np.arctan(np.hypot(dz_dx, dz_dy)))
aspect = np.degrees(np.arctan2(-dz_dx, -dz_dy)) % 360     # arah menurun lereng (azimut dari utara)

# Hillshade (sumber cahaya azimut 315 derajat, elevasi 45 derajat)
zenit, azim = np.radians(45), np.radians(315)
hs = (np.cos(zenit) * np.cos(np.radians(slope)) +
      np.sin(zenit) * np.sin(np.radians(slope)) * np.cos(azim - np.radians(aspect)))
hs = np.clip(hs, 0, 1)

batas = [0, 8, 15, 25, 45, 90]
nama = ["0-8 (landai)", "8-15 (agak miring)", "15-25 (miring)", "25-45 (curam)", ">45 (sangat curam)"]
hist, _ = np.histogram(slope, bins=batas)
persen = hist / hist.sum() * 100

print("ANALISIS KEMIRINGAN LERENG - DATA SINTETIS")
print(f"Kemiringan min / rata-rata / maks : {slope.min():.1f} / {slope.mean():.1f} / {slope.max():.1f} derajat")
for n, p in zip(nama, persen):
    print(f"Kelas {n:<22}: {p:5.1f}% area")
print("Catatan: kelas lereng bersifat contoh klasifikasi, bukan standar baku untuk lokasi tertentu.")

fig, ax = plt.subplots(1, 3, figsize=(15, 4.2))
ax[0].imshow(hs, origin="lower", extent=[1, 299, 1, 199], cmap="gray")
ax[0].set_title("Hillshade")
im1 = ax[1].imshow(slope, origin="lower", extent=[1, 299, 1, 199], cmap="YlOrRd")
ax[1].set_title("Kemiringan lereng"); plt.colorbar(im1, ax=ax[1], label="Derajat", fraction=0.046)
im2 = ax[2].imshow(aspect, origin="lower", extent=[1, 299, 1, 199], cmap="twilight")
ax[2].set_title("Arah hadap lereng"); plt.colorbar(im2, ax=ax[2], label="Azimut (derajat)", fraction=0.046)
for a in ax:
    a.set_xlabel("X lokal (m)"); a.set_ylabel("Y lokal (m)")
plt.tight_layout()
plt.show()
```

**Interpretasi:** Kemiringan lereng berkisar 0,4–32,3° dengan rata-rata 16,6°. Sekitar 66,7% area termasuk kelas miring (15–25°) dan 29,0% agak miring (8–15°), sesuai lereng dasar yang dibuat pada model. Kelas curam (25–45°) hanya 3,1% dan terkonsentrasi di sekitar gawir zona longsoran yang tampak sebagai cincin gelap-terang pada hillshade. Alur lembah lurus di bagian barat juga terlihat jelas pada hillshade. Pembagian kelas di atas hanyalah contoh, sedangkan pada studi nyata klasifikasi mengikuti acuan yang dipakai pada daerah penelitian.

### **4. Perubahan Elevasi Multitemporal (DEM of Difference)**

Data sintetis : dua DEM dari waktu berbeda dengan galat vertikal 0,08 m, disertai simulasi penurunan material pada zona gawir dan penumpukan di kaki lereng.

Untuk menghitung selisih elevasi, menerapkan ambang *Level of Detection* (LoD), dan mengestimasi volume kehilangan serta penambahan material.

```python
# ------------------------------------------------------------
# PERUBAHAN ELEVASI MULTITEMPORAL (DEM OF DIFFERENCE / DoD)
# ------------------------------------------------------------
sigma = 0.08                                           # galat vertikal tiap survei (asumsi, m)
dem_t1 = dtm + rng.normal(0, sigma, dtm.shape)

# Survei kedua: material turun dari zona gawir dan menumpuk di bagian kaki lereng
turun = 1.5 * np.exp(-(((X - 160) / 32) ** 2 + ((Y - 100) / 26) ** 2))
tumpuk = 1.1 * np.exp(-(((X - 245) / 28) ** 2 + ((Y - 100) / 22) ** 2))
dem_t2 = dtm - turun + tumpuk + rng.normal(0, sigma, dtm.shape)

dod = dem_t2 - dem_t1
lod = 1.96 * np.sqrt(2) * sigma                        # Level of Detection (batas perubahan nyata)
sig = np.abs(dod) > lod

erosi = dod[sig & (dod < 0)].sum() * res ** 2
deposisi = dod[sig & (dod > 0)].sum() * res ** 2

print("DEM OF DIFFERENCE - DATA SINTETIS")
print(f"Galat vertikal tiap survei (asumsi) : {sigma:.2f} m")
print(f"Level of Detection (LoD, 95%)       : {lod:.2f} m")
print(f"Perubahan minimum / maksimum        : {dod.min():.2f} / {dod.max():.2f} m")
print(f"Volume kehilangan (erosi)           : {abs(erosi):,.0f} m3")
print(f"Volume penambahan (deposisi)        : {deposisi:,.0f} m3")
print(f"Selisih neto                        : {deposisi + erosi:,.0f} m3")
print("Catatan: nilai berasal dari simulasi; pada data nyata LoD bergantung pada akurasi GCP dan model.")

dod_tampil = np.where(sig, dod, np.nan)
fig, ax = plt.subplots(1, 2, figsize=(13, 4.6))
im = ax[0].imshow(dod, origin="lower", extent=[0, 300, 0, 200], cmap="RdBu", vmin=-1.6, vmax=1.6)
ax[0].set_title("DoD mentah"); plt.colorbar(im, ax=ax[0], label="Perubahan elevasi (m)", fraction=0.046)
im = ax[1].imshow(dod_tampil, origin="lower", extent=[0, 300, 0, 200], cmap="RdBu", vmin=-1.6, vmax=1.6)
ax[1].set_title(f"DoD setelah ambang LoD (|dZ| > {lod:.2f} m)")
plt.colorbar(im, ax=ax[1], label="Perubahan elevasi (m)", fraction=0.046)
for a in ax:
    a.set_xlabel("X lokal (m)"); a.set_ylabel("Y lokal (m)")
plt.tight_layout()
plt.show()
```

**Interpretasi:** Dengan LoD 0,22 m, perubahan yang dianggap nyata tampak sebagai zona kehilangan elevasi (merah) di sekitar X 160 m dan zona penambahan (biru) di sekitar X 245 m, dengan perubahan antara −1,83 m dan +1,33 m. Volume kehilangan 3.746 m³ lebih besar daripada penambahan 2.038 m³, sehingga selisih neto −1.708 m³. Pada simulasi ini ketidakseimbangan tersebut muncul karena zona kehilangan dibuat lebih besar, ditambah bintik-bintik di luar zona utama yang melewati ambang akibat derau (sekitar 5% pada tingkat kepercayaan 95%). Pada data nyata, selisih neto dapat berarti material keluar dari area survei, perubahan kerapatan material, atau galat pengolahan, sehingga tetap perlu dievaluasi dengan akurasi GCP.

### **5. Pengenalan Kelurusan dan Profil Topografi**

Data sintetis : DTM dengan satu alur lembah lurus di bagian barat dan zona longsoran yang dikeluarkan dari analisis.

Untuk mendeteksi kelurusan melalui kelengkungan permukaan, menentukan arahnya, serta membandingkan DSM dan DTM pada satu profil.

```python
# ------------------------------------------------------------
# PENGENALAN KELURUSAN & PROFIL TOPOGRAFI
# ------------------------------------------------------------
# Kelurusan lembah dicari dari kelengkungan permukaan (turunan kedua DTM) di luar zona longsoran
zona_longsor = (((Xc - 165) / 70) ** 2 + ((Yc - 100) / 55) ** 2) < 1
kelengkungan = np.gradient(dz_dx, res, axis=1) + np.gradient(dz_dy, res, axis=0)
ambang = np.percentile(kelengkungan[~zona_longsor], 99)
kandidat = (kelengkungan > ambang) & (~zona_longsor)

px, py = Xc[kandidat], Yc[kandidat]
pusat = np.array([px.mean(), py.mean()])
cov = np.cov(np.vstack([px - pusat[0], py - pusat[1]]))
nilai, vektor = np.linalg.eigh(cov)
arah = vektor[:, np.argmax(nilai)]                         # sumbu utama sebaran titik
strike = np.degrees(np.arctan2(arah[0], arah[1])) % 180    # azimut kelurusan (0-180)

print("PENGENALAN KELURUSAN - DATA SINTETIS")
print(f"Ambang kelengkungan (persentil 99) : {ambang:.3f} 1/m")
print(f"Jumlah piksel kandidat           : {kandidat.sum()}")
print(f"Azimut kelurusan terdeteksi      : N {strike:.0f} E")
print("Catatan: kelurusan pada model dibuat sengaja; pada data nyata perlu dicek dengan geologi lapangan.")

# Profil memotong zona longsoran pada Y = 100 m
j = 100
fig, ax = plt.subplots(1, 2, figsize=(14, 4.6))
ax[0].imshow(hs, origin="lower", extent=[1, 299, 1, 199], cmap="gray")
ax[0].scatter(px, py, s=4, color="tab:red", label="Kandidat kelurusan")
t = np.array([-60, 60])
ax[0].plot(pusat[0] + t * arah[0], pusat[1] + t * arah[1], color="cyan", lw=1.5, label="Garis kelurusan")
ax[0].axhline(j, color="yellow", ls="--", lw=1, label="Garis profil (Y = 100 m)")
ax[0].set_xlim(0, 300); ax[0].set_ylim(0, 200)
ax[0].set_title("Hillshade dan kelurusan"); ax[0].legend(loc="upper right", fontsize=8)
ax[0].set_xlabel("X lokal (m)"); ax[0].set_ylabel("Y lokal (m)")

ax[1].fill_between(x, dtm[j], dsm[j], color="tab:green", alpha=0.35, label="Vegetasi (DSM - DTM)")
ax[1].plot(x, dsm[j], color="tab:green", lw=1, label="DSM")
ax[1].plot(x, dtm[j], color="saddlebrown", lw=2, label="DTM")
ax[1].set_xlabel("Jarak X (m)"); ax[1].set_ylabel("Elevasi (m)")
ax[1].set_title("Profil topografi memotong zona longsoran"); ax[1].legend(); ax[1].grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

**Interpretasi:** Sebanyak 470 piksel dengan kelengkungan di atas ambang 0,064 1/m tersusun lurus dan menghasilkan azimut N 117° E, sesuai arah alur yang sengaja dibuat pada model. Hal ini memperlihatkan bahwa kelurusan dapat dikenali secara objektif dari turunan DTM, bukan hanya dari pengamatan visual. Pada data nyata, kelurusan baru dapat dikaitkan dengan struktur geologi setelah dicocokkan dengan peta geologi dan pengamatan lapangan, karena alur lurus juga dapat terbentuk oleh erosi. Profil di sebelah kanan menegaskan bahwa vegetasi menaikkan DSM di atas DTM, sehingga DTM lebih sesuai untuk membaca bentuk permukaan tanah.
